# BroadCast Variable

```python
from pyspark.sql import functions as F

data = [("101","John","USA","CA"),
        ("102","Ray","USA","NY"),
        ("103","Ron","USA","NY"),
        ("104","Smith","USA","FL")]

df = spark.createDataFrame(data, ["id", "name", "country", "state"])

EmpEmails = {"101":"john@gmail.com", "102":"ray@gmail.com",
             "103":"ron@gmail.com", "104":"smith@gmail.com"}

emails_df = spark.createDataFrame(list(EmpEmails.items()), ["id", "email"])

result = df.join(F.broadcast(emails_df), "id", "left") \
           .select("id", "name", "country", "state", "email")

display(result)
print([tuple(r) for r in result.collect()])
```
