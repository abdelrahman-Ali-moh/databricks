```python
import requests

url = "https://health.data.ny.gov/api/v3/views/xdss-u53e/query.csv"
path = "/Volumes/workspace/default/my_volume/ny_hospitals.csv"

response = requests.get(url, timeout=120)
response.raise_for_status()

with open(path, "wb") as f:
    f.write(response.content)

baby = spark.read.csv(path, header=True, inferSchema=True)

# أسماء الأعمدة فيها مسافات ورموز، بنظّفها عشان saveAsTable
import re
baby = baby.toDF(*[re.sub(r"[^0-9a-zA-Z_]", "_", c) for c in baby.columns])

baby.write.mode("overwrite").saveAsTable("workspace.default.baby")
display(baby)
```
