# Second type

```python
from pyspark.sql.functions import col
from pyspark.sql.types import StringType,IntegerType,DoubleType
```

```python
df3=df.withColumn("id",df.id.cast(IntegerType())).withColumn("salary",df.salary.cast(DoubleType()))
```

```python
df3.display()
```

# Third Type

```python
df4=df.withColumn("id",col("id").cast("integer")).withColumn("salary",col("salary").cast("double"))
```

```python
df4.printSchema()
```

# Casting using selectExpr method

```python
df2=df.selectExpr("cast(id as integer) id", "name", "country", "state", "email", "cast(salary as double) salary")
df2.printSchema()
```

# SQL Style Casting

```python
df.createOrReplaceTempView("CastExample")
df4 = spark.sql("SELECT boolean(id) AS id_bool FROM CastExample")
df4.printSchema()
```
