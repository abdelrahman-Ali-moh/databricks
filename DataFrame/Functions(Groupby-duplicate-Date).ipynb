```python
data = [
    (1, 2000, "it", "abdo","2022-07-10 15:12:13"),
    (2, 3000, "ACC", "Gomma","2025-08-01 17:10:10"),
    (2, 3000, "ACC", "Gomma","2026-01-01 18:10:10"),
    (3, 10000, "ACC", "Ali","2026-01-01 18:10:10"),
    (4, 20000, "it", "mohamed","2026-01-01 18:10:10"),
    (5,1500 , "ACC", "ahmed","2026-01-01 18:10:10")
]

columns = ["id", "salary", "Dept", "name","hire_date"]

df = spark.createDataFrame(data, columns)
df.show()
```

```python
from pyspark.sql.functions import *

df.groupBy("Dept").sum("salary").show(truncate=False)
```

```python
df.groupBy("Dept").avg("salary").show(truncate=False)
```

# Drop Duplicated

```python
df.drop_duplicates().show()
```

```python
df.drop_duplicates(["name"]).show()
```

# Date

```python
df=(df.select("hire_date").withColumn("next_month",add_months("hire_date",1)))
df.show()
```

# Current Date

```python
df=(df.select("hire_date").withColumn("current_date",current_date())).show()
```
