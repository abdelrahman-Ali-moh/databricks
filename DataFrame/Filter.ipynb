```python
df=spark.read.format("csv").option("header",True).load("/Workspace/Users/abdelrahman2556250@gmail.com/databricks/employees.csv")
```

# Equals

```python
df.filter(df.salary<=50000).show()
```

# Not Equals

```python
df.filter(df.salary!=4000).show()
```

# Filter on Multiple Fields

```python
df.filter((df.country == "Egypt")&(df.salary>1000)).show()
```

# Text Filters

```python
df.filter(df.name.startswith("A")).show()
```

```python
df.filter(df.name.endswith("A")).show()
```

# Null Filters

```python
df.filter(df.country.isNull()).show()
df.filter(df.country.isNotNull()).show()
```

#SQL Way Filters

```python
df.createOrReplaceTempView("DATA")
spark.sql("select * from DATA where State == 'CA'").show()
```
