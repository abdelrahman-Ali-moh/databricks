#Create a new Schema

```python
from pyspark.sql.types import StructType

Schema_def=StructType()
Schema_def.add('id', "string", True),
Schema_def.add('name', "string", True),
Schema_def.add('country', "string", True),
Schema_def.add('state', "string", True),
Schema_def.add('email', "string", True),
Schema_def.add('salary', "integer", True)
```

```python
df=spark.read.format("csv").option("header",True).schema(Schema_def).load("/Workspace/Users/abdelrahman2556250@gmail.com/databricks/employees.csv")
```

```python
display(df)
```

# Select Column

```python
df2=df.select(df.id,df.email,df.name)
display(df2)
```

```python
df3=df.select(df['id'],df['country'])
display(df3)
```

```python
from pyspark.sql.functions import col

df4=df.select(col('id'),col('email'))
display(df4)
```

# WriteColumn

```python
df5=df.withColumn('newcolumn',df.salary*0.5)
```

```python
display(df5)
```

# Create Lit (عبارة عن عمل عمود ووضع كلمة للتكرار )

```python
from pyspark.sql.functions import lit

df6=df.withColumn("lit attrition",lit('Yes'))
display(df6)
```

```python
from pyspark.sql.functions import concat_ws

df7=df.withColumn('email',concat_ws('@',df.name,df.email))
display(df7)
```

# Rename Columns

```python
df8=df.withColumnRenamed('id','newid')
display(df8)
```

#DropFields

```python
df9=df8.drop("newid").show()
```

# when & otherwise

```python
from pyspark.sql.functions import when, col

df10 = when(col("salary") < 5000, "low").when(col("salary") >= 50000, "medium").otherwise("high").alias("salary_level")
display(df10)
```

```python
df=df.withColumn('salary_level',df10)
df.show()
```
